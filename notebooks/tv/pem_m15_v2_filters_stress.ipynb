# PEM M15 v2 — Filter + Stress Test

Notebook ini menguji kandidat PEM M15 v2 berdasarkan hasil edge localization.

Varian:
- baseline
- block NY Overlap
- mid/high volatility only
- block NY Overlap + mid/high volatility
- Asia only
- Asia + NY Late
- Asia + mid/high volatility

Stress test:
- delay entry
- spread tambahan
- slippage tambahan
- TP haircut


```python
from xauusd_pem_vwap_research import load_ohlcv, resample_ohlcv
from pem_m15_v2_filters_stress import prepare_sig_with_filters, apply_variant, run_variants, summarize_variant_runs, run_stress_tests_for_variant, plot_bar, save_outputs
```

```python
DATA_PATH = 'xauusd_m1.csv'  # ganti sesuai file kamu
df_m1 = load_ohlcv(DATA_PATH)
df_m15 = resample_ohlcv(df_m1, '15min')
print('Rows M15:', len(df_m15))
```

```python
variant_runs = run_variants(df_m15)
variant_summary = summarize_variant_runs(variant_runs)
variant_summary
```

```python
plot_bar(variant_summary, 'variant', 'expectancy_r', 'PEM M15 v2 Variants — Expectancy')
plot_bar(variant_summary, 'variant', 'profit_factor', 'PEM M15 v2 Variants — Profit Factor')
```

```python
CANDIDATE_VARIANT = 'block_ny_overlap_mid_high_vol'  # ubah jika perlu
sig = prepare_sig_with_filters(df_m15)
sig_variant = apply_variant(sig, CANDIDATE_VARIANT)
stress_df = run_stress_tests_for_variant(sig_variant, CANDIDATE_VARIANT)
stress_df
```

```python
plot_bar(stress_df, 'scenario', 'expectancy_r', f'{CANDIDATE_VARIANT} — Stress Expectancy')
plot_bar(stress_df, 'scenario', 'profit_factor', f'{CANDIDATE_VARIANT} — Stress Profit Factor')
```

```python
OUT = 'outputs_pem_m15_v2'
save_outputs(OUT, variant_summary, stress_df, variant_runs)
print('saved to', OUT)
```

## Cara baca hasil

Cari varian yang:
- expectancy lebih tinggi dari baseline
- PF membaik
- max DD mengecil
- trade count tidak turun terlalu ekstrem

Untuk stress test, kandidat yang baik tidak langsung runtuh saat delay +1, spread kecil, atau TP haircut ringan.
