# Edge Localization — PEM M15

Notebook ini dipakai untuk mencari **di mana edge PEM M15 muncul**.

Analisis:
- session breakdown
- hour breakdown
- volatility bucket breakdown
- trend slope bucket breakdown
- filter experiments


```python
from xauusd_pem_vwap_research import load_ohlcv, resample_ohlcv, backtest_pem
from pem_edge_localization import attach_pem_context, group_metrics, filter_experiments, experiment_summary, plot_group_bar, save_edge_outputs
```

```python
DATA_PATH = 'xauusd_m1.csv'  # ganti sesuai file kamu
df_m1 = load_ohlcv(DATA_PATH)
df_m15 = resample_ohlcv(df_m1, '15min')
print('Rows M15:', len(df_m15))
```

```python
pem_m15 = backtest_pem(df_m15, timeframe='M15', rr_target=2.0, timeout_bars=12)
pem_ctx = attach_pem_context(df_m15, pem_m15)
pem_ctx.head()
```

```python
by_session = group_metrics(pem_ctx, 'session')
by_hour = group_metrics(pem_ctx, 'hour')
by_vol = group_metrics(pem_ctx, 'vol_bucket')
by_trend = group_metrics(pem_ctx, 'trend_bucket')
by_session, by_vol, by_trend
```

```python
plot_group_bar(by_session, 'session', 'expectancy_r', 'PEM M15 Expectancy by Session')
plot_group_bar(by_hour, 'hour', 'expectancy_r', 'PEM M15 Expectancy by Hour')
plot_group_bar(by_vol, 'vol_bucket', 'expectancy_r', 'PEM M15 Expectancy by Vol Bucket')
plot_group_bar(by_trend, 'trend_bucket', 'expectancy_r', 'PEM M15 Expectancy by Trend Bucket')
```

```python
experiments = filter_experiments(df_m15)
exp_summary = experiment_summary(experiments)
exp_summary
```

```python
plot_group_bar(exp_summary, 'experiment', 'expectancy_r', 'PEM M15 Filter Experiment Expectancy')
plot_group_bar(exp_summary, 'experiment', 'profit_factor', 'PEM M15 Filter Experiment Profit Factor')
```

```python
OUT = 'outputs_pem_edge'
save_edge_outputs(OUT, pem_ctx, by_session, by_hour, by_vol, by_trend, exp_summary)
print('saved to', OUT)
```
